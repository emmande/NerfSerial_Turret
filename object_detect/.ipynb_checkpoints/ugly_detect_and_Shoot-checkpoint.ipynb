## Saving Face Encodings to use later for face similarity

```python
import face_recognition as fr # To recognize faces
import numpy as np # To handle all lists/arrays
import cv2 # To capture webcam footage
import os # To handle all matters relating to folders, paths, image/file names, etc.
import pickle

# Path to folder containing all known faces. The 'known' folder needs all known faces, 
# where the image\file name is the name of the person to be recognized.
faces_path = "C:\\Users\\User\\Documents\\NerfSerial_Turret\\object_detect\\FacialRecognition\\fr"

# Function to get face names, as well as face encodings
def get_face_encodings():
    face_names = os.listdir(faces_path)
    face_encodings = []

    # For loop to retrieve all face encodings and store them in a list.
    # Below loop also gets the names of people and removes ".jpg", and stores
    # the names in a list
    for i, name in enumerate(face_names):
        face = fr.load_image_file(f"{faces_path}\\{name}")
        face_encodings.append(fr.face_encodings(face)[0])

        face_names[i] = name.split(".")[0][:-1] # To remove ".jpg" or any other image extension
    
    return face_encodings, face_names

# Retrieving face encodings and storing them in the face_encodings variable, along with the names
face_encodings, face_names = get_face_encodings()


# Add this to picled file

data ={"encodings": face_encodings, "names": face_names}#use pickle to save data into a file for later use
f = open("C:\\Users\\User\\Documents\\NerfSerial_Turret\\object_detect\\FacialRecognition\\face_enc_panget_pogi", "wb")
f.write(pickle.dumps(data))#to open file in write mode
f.close()#to close file
```

## Identify, Track and Shoot Faces 

```python
import face_recognition as fr # To recognize faces
import numpy as np # To handle all lists/arrays
import cv2 # To capture webcam footage
import os # To handle all matters relating to folders, paths, image/file names, etc.
import pickle
import serial
import time


try:
    serialcomm = serial.Serial(    
            # port="COM3", baudrate=9600 # for arduino uno
    port="COM7", baudrate=74880 #115200 # for Adafruit Metro
    # port="/dev/ttyACM0", baudrate=9600
    , bytesize=8, timeout=2, stopbits=serial.STOPBITS_ONE)
    
except Exception:
    serialcomm.close()
    serialcomm = serial.Serial(    
    # port="COM3", baudrate=9600 # for arduino uno
    port="COM7", baudrate=74880 #115200 # for Adafruit Metro
    # port="/dev/ttyACM0", baudrate=9600
    , bytesize=8, timeout=2, stopbits=serial.STOPBITS_ONE)



# serialcomm.timeout = 1

def upturn():
    i="u"    
    serialcomm.write(i.encode())

def downturn():
    i="d"    
    serialcomm.write(i.encode())

def leftturn():
    i="l"    
    serialcomm.write(i.encode())

def rightturn():
    i="r"    
    serialcomm.write(i.encode())
    
def fire():
    i="f"    
    # time.sleep(0.2)
    serialcomm.write(i.encode())

def m_stop():
    i="s"    
    # time.sleep(0.2)
    serialcomm.write(i.encode())


# load the known faces and embeddings saved in last file
data = pickle.loads(open('C:\\Users\\User\\Documents\\NerfSerial_Turret\\object_detect\\FacialRecognition\\face_enc_panget_pogi', "rb").read())


face_encodings = data["encodings"]
face_names = data["names"]


# Reference to webcam
video = cv2.VideoCapture(0)

# Setting variable which will be used to scale size of image
scl = 4


button_fire = False
button_guapo = False
button_ganda = False
button_panget = False

fire_count=1

def click_button(event, x, y, flags, params):
    global button_guapo, button_ganda, button_panget, button_fire
    if event == cv2.EVENT_LBUTTONDOWN:
#         print(x,y)
        poly_guapo = np.array([[(20,20),(220,20),(220,70),(20,70)]])
        poly_ganda = np.array([[(20,80),(220,80),(220,130),(20,130)]])
        poly_panget = np.array([[(20,140),(220,140),(220,190),(20,190)]])
        poly_fire = np.array([[(20,200),(220,140),(220,250),(20,250)]])

        # print(poly)
        is_inside_guapo =  cv2.pointPolygonTest(poly_guapo, (x,y), False)
        is_inside_ganda =  cv2.pointPolygonTest(poly_ganda, (x,y), False)
        is_inside_panget =  cv2.pointPolygonTest(poly_panget, (x,y), False)
        is_inside_fire =  cv2.pointPolygonTest(poly_fire, (x,y), False)
        
        if is_inside_guapo > 0:
            if button_guapo == False:
                button_guapo=True
                button_ganda = False
                button_panget = False
            else:
                button_guapo = False
                
                    
        if is_inside_ganda > 0:       
            if button_ganda == False:
                button_ganda=True
                button_guapo=False
                button_panget = False
                
            else:
                button_ganda = False
                
        if is_inside_panget > 0:       
            if button_panget == False:
                button_panget=True
                button_guapo=False
                button_ganda = False
            else:
                button_panget = False
                
        if is_inside_fire > 0: 
#                 reset fire to 0 and will fire again
            # fire_count=0
            if button_fire == False:
                button_fire=True
                fire_count=0
            else:
                button_fire = False
    


dicSelect = {False:"OFF", True:"ON"}
        
    
def track_target(x,y,poly_Frame_center_bound,fire_count=0):
    
    x_center = int(x+w//2)
    y_center = int(y+h//2)
    # print(poly_Frame_center_bound)
    # print(x_center,y_center)
    isCrosshairInCenter = cv2.pointPolygonTest(poly_Frame_center_bound, (x_center,y_center), False)
    # print(isCrosshairInCenter)
    # print(poly_Frame_center_bound[0][0][0])
    # print(poly_Frame_center_bound[0][0][1])
    if isCrosshairInCenter != 1:
        m_stop()
        if    x_center < poly_Frame_center_bound[0][0][0] :
            leftturn()
            # print(">>>")
        elif    x_center > poly_Frame_center_bound[0][1][0] :
            rightturn()
            # print("<<<")
                            
        elif  y_center < poly_Frame_center_bound[0][0][1]:#ver_ref+h//10
            downturn()
            # print("up")
        elif   y_center > poly_Frame_center_bound[0][1][1]:
            upturn()
            # print("down")
    
    elif isCrosshairInCenter ==1 and button_fire == True:
        # fire()
        if fire_count < 6:
            fire()
            fire_count=fire_count+1
        else:
            m_stop()
        # isCrosshairInCenter=0 # reset to not in crosshairs
    return fire_count


#create Window
cv2.namedWindow("Frame")
cv2.setMouseCallback("Frame", click_button)

# Continuously capturing webcam footage
while True:
    success, image = video.read()
    image = cv2.resize(image, (0, 0),fx=2,fy=2)
    # Making current frame smaller so program runs faster
    resized_image = cv2.resize(image, (int(image.shape[1]/scl), int(image.shape[0]/scl)))
    #     resize frame to something bigger

    # check for center of frame
    rad = 30
    (h_frm, w_frm) = image.shape[:2] #w:image-width and h:image-heightq
    cv2.circle(image, (w_frm//2, h_frm//2), rad, (0, 255, 0), 1) 
    
    hor_ref = w_frm//2 # Horizontal reference
    ver_ref = h_frm//2 # vertical reference

        #will use this polygon as boundary to check if center of cross hair is within this
    hor_start= max(hor_ref-rad,0)
    ver_start=max(ver_ref-rad,0)
    hor_end=max(hor_ref+rad,0)
    ver_end=max(ver_ref+rad,0)
    cv2.line(image, (hor_start,ver_start), (hor_end, ver_start), (0,0,30), 2)
    cv2.line(image, (hor_start,ver_end), (hor_end, ver_end), (0,0,30), 2)
    poly_Frame_center_bound = np.array([[(hor_start,ver_start),(hor_end,ver_start),
                                         (hor_end,ver_end),(hor_start,ver_end)]])


    
    # Converting current frame to RGB, since that's what the face recognition module uses
    rgb_image = cv2.cvtColor(resized_image, cv2.COLOR_BGR2RGB)

    # Retrieving face location coordinates and unknown encodings
    face_locations = fr.face_locations(rgb_image)
    unknown_encodings = fr.face_encodings(rgb_image, face_locations)

    # Iterating through each encoding, as well as the face's location
    for face_encoding, face_location in zip(unknown_encodings, face_locations):
        # Comparing known faces with unknown faces
        result = fr.compare_faces(face_encodings, face_encoding, 0.4)

        # Getting correct name if a match was found
        if True in result:
            name = face_names[result.index(True)]

            # Setting coordinates for face location
            top, right, bottom, left = face_location

            x = left*scl
            y = top*scl
            h = (bottom-top)*scl
            w= (right-left)*scl
            
            # Drawing rectangle around face
            cv2.rectangle(image, (left*scl, top*scl), (right*scl, bottom*scl), (0, 0, 255), 2)
            #cv2.rectangle(frame, (x,y), (x+w, y+h), (200,0,50), 3)

            
            # Setting font, as well as displaying text of name
            font = cv2.FONT_HERSHEY_DUPLEX
            cv2.putText(image, name, (left*scl, bottom*scl + 20), font, 0.8, (255, 255, 255), 1)
            
            if name == 'guapo' and button_guapo == True:
                # draw cross hairs
                cv2.line(image, (x+w//3,y+h//2), (x+2*w//3, y+h//2), (0,0,255), 2)
                cv2.line(image , (x+w//2,y+h//3), (x+w//2, y+2*h//3), (0,0,255), 2)
                
                fire_count=track_target(x,y,poly_Frame_center_bound,fire_count)

            if name == 'ganda' and button_ganda == True:
                # draw cross hairs
                cv2.line(image, (x+w//3,y+h//2), (x+2*w//3, y+h//2), (0,0,255), 2)
                cv2.line(image , (x+w//2,y+h//3), (x+w//2, y+2*h//3), (0,0,255), 2)
                
                fire_count=track_target(x,y,poly_Frame_center_bound,fire_count)

            if name == 'panget' and button_panget == True:
                # draw cross hairs
                cv2.line(image, (x+w//3,y+h//2), (x+2*w//3, y+h//2), (0,0,255), 2)
                cv2.line(image , (x+w//2,y+h//3), (x+w//2, y+2*h//3), (0,0,255), 2)
                
                fire_count=track_target(x,y,poly_Frame_center_bound,fire_count)


    #     #create buttons
    cv2.rectangle(image, (20,20), (220,70), (0,0,200), -1)
    cv2.putText(image, f"Guapo {dicSelect.get(button_guapo)}", (30,60), cv2.FONT_HERSHEY_PLAIN, 2, (255,255,255), 3)
    cv2.rectangle(image, (20,80), (200,130), (0,0,180), -1)
    cv2.putText(image, f"Ganda {dicSelect.get(button_ganda)}", (30,120), cv2.FONT_HERSHEY_PLAIN, 2, (255,255,255), 3)
    cv2.rectangle(image, (20,140), (220,190), (0,0,200), -1)
    cv2.putText(image, f"Panget {dicSelect.get(button_panget)}", (30,180), cv2.FONT_HERSHEY_PLAIN, 2, (255,255,255), 3)
    cv2.rectangle(image, (20,200), (150,240), (200,0,200), -1)
    cv2.putText(image, f"FIRE {dicSelect.get(button_fire)}", (30,230), cv2.FONT_HERSHEY_PLAIN, 2, (255,255,255), 3)
        

    # Displaying final image on the screen
    cv2.imshow("Frame", image)
    if cv2.waitKey(1) == ord('q'):
        m_stop()
        break
video.release()
cv2.destroyAllWindows()
serialcomm.close()
```
